```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('MOCK_DATA.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
sns.histplot(df['gender'], kde = True)
plt.title('Distribution of gender')
plt.tight_layout()
plt.show()
```

```python
df_num = df.select_dtypes(include = 'number')

sns.heatmap(df_num.corr(), annot = True, cmap = 'coolwarm')
plt.title('Correlation Matrix')
plt.figure(figsize = (10, 10))
plt.tight_layout()
plt.show()
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
df.info()
```

```python
sns.scatterplot(data = df, x = 'sepal_length', y = 'sepal_width', hue = 'species')
plt.title('scatter plot between sepal length and sepal width')
plt.tight_layout()
plt.legend()
plt.show()
```

```python
sns.countplot(data = df, x = 'petal_width', hue = 'species')
plt.title('count plot between petal width and species')
plt.tight_layout()
plt.legend()
plt.show()
```

```python
sns.boxplot(data = df, x = 'species', y = 'petal_length')
plt.title('box plot between species and petal length')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

# variable types
employee_name = 'Alice'
employee_age = 29
employee_height = 5.4
is_student = False

print('employee info:', employee_name, '| age', employee_age, '| height', employee_height, '| student', is_student)

# array
monthly_sales = [250, 300, 400, 150, 500]
sales_array = np.array(monthly_sales)

print('\n Sales Analysis:')
print('total sales:', sales_array.sum())
print('average sales:', sales_array.mean())
print('highest sales:', sales_array.max())

data = pd.DataFrame({
    'Name' : ['Alice','Bob','charles'],
    'Dep' : ['HR','Finance','Marketing'],
    'Salary': [30000,40000,60000]
})

data['Department_lower'] = data['Dep'].str.lower()

high_earners = data['Salary'] > 50000
filtered_data = data[high_earners]

print('\nFiltered Data:')
print(filtered_data)

print('\n Employee Data:\n', data)
print('\n High Earners:\n', filtered_data)
```

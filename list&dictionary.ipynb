```python
# list can also refer to an array of data
# list could be a collection of numerical value or strings values
```

```python
# define an array (list) of data
fruits = ['apple','cherry','banana']
```

```python
# indexing
# gía trị đầu tiên sẽ là index = 0
print(fruits[1]) # cherry là index = 1
```

```python
# change the item
fruits[2] = 'fig'
print(fruits[2])
```

```python
# append - add new item
fruits.append('grapes')
print(fruits[3])
```

```python
# remove item
fruits.remove('cherry')
print(fruits[1])
```

```python
# looping through a list
# print all items in list
for fruit in fruits:
  print(fruit)
```

```python
# dictionaries store data  in key value pairs
# defined using curly braces or third bracket
```

```python
# create dictionary - looks like an observation
student = {'name':'Alice','age': 21, 'grade': 'A', 'nationality':'USA'}
```

```python
print(student['name'])
```

```python
# change the value
student['age'] = 25
print(student['age'])
```

```python
# adding new key pair value
student['subject'] = 'Math'
print(student['subject'])
```

```python
# remove a key pair value
del student['grade']
```

```python
# looping through a dictionary
for key, value in student.items():
  print(f":{key}: {value})")

# f" là chuỗi định dạng f-string: f-string là cách viết chuỗi trong Python cho phép chèn biến trực tiếp vào chuỗi, vừa ngắn, vừa dễ đọc, lại chạy nhanh.
```

```python
# test lại cái looping
for cut, phan in student.items():
  print(f"{cut} : {phan}")
```

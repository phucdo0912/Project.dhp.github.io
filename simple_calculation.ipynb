```python
# list
number = [2,4,6,8,10]

print(number)
```

```python
# loop for list
for num in number:
  print(num)
```

```python
for num in number:
  print(num+1)
  # có thể làm tương tự với các phép toán học khác)
```

```python
# store iteration in a specific new list
doubled = []
for num in number:
  doubled.append(num * 2)

print('doubled list:', doubled)
```

```python
# cách ngắn hơn - one line iteration
doubled = [num * 2 for num in number]
print(doubled)
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data1 = {
    'Name': ['Alice', 'Bob', 'Charles', 'David', 'Eva'],
    'Age': [24, np.nan, 30, 22, np.nan],
    'Gender':['F','M', np.nan,'M', 'F'],
    "Score": [85,90, np.nan, 66, 76]
}

df = pd.DataFrame(data1)
df
```

```python
# check for missing values
df.isnull().sum()
```

```python
print(df.isnull())
```

```python
df_drop_any = df.dropna() # xóa dòng nếu dòng có  NaN - chỉ cần dòng có NaN là xóa
```

```python
# drop rows where values are missing
df_drop_all = df.dropna(how = 'all') # xóa dòng nếu tất cả các giá trị của dòng đều là NaN
```

```python
# drop column
df_drop_column = df.dropna(axis = 1)
```

```python
# filling missing data
```

```python
# fil with a specific value
df_fill_constant = df.fillna(0)
```

```python
# fill with mean/median/mode
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Score'] = df['Score'].fillna(df['Score'].median())
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
print(df)
```

```python
# fill foward
df_ffill = df.ffill()
```

```python
# fil backward
df_bfill = df.bfill()
```

```python
# interpolating with missing data - nội suy - numerical
df_interpolated = df.interpolate() # old version
```

```python
# new version
# vì trong các phiên bản mới interpolate chỉ dùng cho numeric data nên phải dùng selec_dtypes để lựa num data
df_interpolated_new = df.select_dtypes(include = 'number').interpolate()
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charles', 'David', 'Eva'],
    'Age': [2,37,30,22,55],
    'Gender':['F','M', 'M','M', 'F'],
    "Salary": [50000,60000,5500,80000,45000]
}

df1 = pd.DataFrame(data)
```

```python
print(df1)
```

```python
condition = df1['Age'] >= 30
```

```python
print(condition)
```

```python
# filter dataframe
print(df1[condition])
df1[condition]
```

```python
filtered_df = df1[(df1['Age'] > 20) & (df1['Gender'] == "M")]
print(filtered_df)
```

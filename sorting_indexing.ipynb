```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Name':['Alice','Bob','Charles','David'],
    'Age':[25,56,33,24],
    'Score':[85,90,78,88],
    'Country':['USA', 'Canada', 'UK', 'UAE']
}

df = pd.DataFrame(data)
print(df)
```

```python

# sorting
df_sorted_age = df.sort_values(by = "Score")
print(df_sorted_age)
```

```python
# sort multiple values
# sort nhiều giá trị thì để trong ngoặc vuông
df_sorted = df.sort_values(by =['Age','Score'], ascending = [False, True])
df_sorted
```

```python
# sort by index
df_sorted_index = df.sort_index(ascending=False)
print(df_sorted_index)
```

```python
# indexing
```

```python
# đổi index thành dạng label lấy tham chiếu từ cột Name (vd: name: alice => index: alice)
df.set_index('Name', inplace = True)

# dùng df.loc để lấy index dạng label
print(df.loc['Alice'])
```

```python
# index by index of rows
print(df.iloc[3])
```

```python
# access specific rows and columns
print(df.loc[['Charles','Bob'], ['Age','Score','Country']])
```

```python
# reset index back to integer
df_reset = df.reset_index()
print(df_reset)
```

```python
# creating new columns
```

```python
data = {
    'Name':['Alice','Bob','Charles','David'],
    'Math':[88,56,92,88],
    'Science':[90,82,85,95]
}

df = pd.DataFrame(data)
print(df)
```

```python
# new column
df["Average"] = (df['Math'] + df['Science'])/2
print(df)
```

```python
df['Result'] = df['Average'].apply(lambda x: "Pass" if x > 85 else 'Fail')
print(df)
```

```python
df['Year'] = 2025
print(df)
```

```python
def grade(score): # define một hàm để apply bên dưới
  if score >= 90:
    return 'A'
  elif score >=80:
    return 'B'
  else:
    return 'C'

df['Grade'] = df['Average'].apply(grade) # apply hàm được define ở trên
print(df)
```

```python
# lambda, np.where and map
# lambda: inline function used with apply for custom logic
# np.where: numpy function for vectorized conditional operation
# map: function to map values in a common to new values, often used for lists
```

```python
# lambda
df['Math_level'] = df['Math'].apply(lambda x: 'High' if x>80 else 'Low')
print(df)
```

```python
# np.where
# hoạt động như hàm if trong excel (logical, value if true, value if false)
df['Science_level'] = np.where(df['Science'] >=90, 'Top', 'Average')
print(df)
```

```python
# map
# có sẵn grade phía trên

grade_map = {
    'A': 'Excellent',
    'B': 'Good',
    'C': 'Needs Improvement'
}

df['Grade_description'] = df['Grade'].map(grade_map)
print(df)
```

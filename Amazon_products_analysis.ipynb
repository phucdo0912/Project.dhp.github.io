```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_rows', 200)
```

```python
# load dataset
df = pd.read_csv('amazon.csv')
df.head()
```

Data cleaning

```python
# number of products
df['product_id'].nunique()
```

```python
# drop duplicate product_id
df = df.drop_duplicates(subset = 'product_id')
df
```

```python
# cleaning unwanted symbols in pice columns
df['discounted_price'] = df['discounted_price'].str.replace('₹','', regex = False).str.replace(',','', regex = False)
df['actual_price'] = df['actual_price'].str.replace('₹','', regex = False).str.replace(',','', regex = False)

df['discount_percentage'] = df['discount_percentage'].str.replace('%','', regex = False).str.replace(',','', regex = False)
df['rating_count'] = (df['rating_count'].str.replace('%','',regex=False).str.replace(',','',regex=False))
```

```python
# convert columns into suitable type
df['discounted_price'] = pd.to_numeric(df['discounted_price'], errors = 'coerce')
df['actual_price'] = pd.to_numeric(df['actual_price'], errors = 'coerce')
df['discount_percentage']=pd.to_numeric(df['discount_percentage'], errors='coerce') / 100
df['rating']=pd.to_numeric(df['rating'], errors='coerce')
df['rating_count']=pd.to_numeric(df['rating_count'], errors='coerce')

df = df.dropna(subset = ['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count'])
```

```python
# check for missing values
df.isnull().sum()
```

Exploratory data analysis (EDA)

```python
# top rated categories based on rating & rating count => weighted average
top_categories = df.groupby('category').apply(lambda x: (x['rating'] * x['rating_count']).sum() / x['rating_count'].sum(), include_groups = False).sort_values(ascending = False).head(50)

top_categories
```

```python
# top 10 discounted categories
df.groupby('category')['discount_percentage'].mean().sort_values(ascending = False).head(10)
```

```python
# top 10 reviewed products
df.sort_values('rating_count', ascending = False).head(10)
```

```python
# most reviewed categories
df.groupby('category')['rating_count'].sum().sort_values(ascending = False).head(10)
```

```python
# discount percentage by ratings
df.groupby('rating')['discount_percentage'].mean().sort_values(ascending = False)
```

```python
# correlation matrix
plt.figure(figsize = (8,6))
sns.heatmap(data = df.select_dtypes(include = 'number').corr(), annot = True, cmap = 'seismic')
plt.title('Correlation matrix')
plt.tight_layout
plt.show()
```

```python
# discounted price vs rating
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='discounted_price', y='rating', color='#1800ad', alpha=0.3, s = 200)
plt.title('discounted price vs rating')
plt.tight_layout()
plt.show()
```

```python
# price vs number of rates
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='discounted_price', y='rating_count', color='#d10101', alpha=0.3, s = 200)
plt.title('discounted price vs number of rates')
plt.tight_layout()
plt.show()
```

Feature engineering

```python
# consumer surplus
df['saving'] = df['actual_price'] - df['discounted_price']
# the length of review
df['review_length'] = df['review_content'].astype(str).str.len()
```

```python
plt.figure(figsize = (8,6))
sns.boxplot(data = df, y = 'review_length', x = 'rating', color = '#1800ad')
plt.title('relationship between rating and review length')
plt.tight_layout()
plt.show()
```

```python
# calculate scoring value for products
df['product_score'] = df['rating'] * df['discount_percentage'] * np.log1p(df['rating_count'])
df.sort_values('product_score', ascending = False).head(10)
```

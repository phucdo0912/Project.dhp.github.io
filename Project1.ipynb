```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset = 'age')
df
```

```python
df['survived'].value_counts()
```

```python
sns.countplot(data = df, x = 'survived', hue = 'sex')
plt.title('Survial count by gender')
plt.show()
```

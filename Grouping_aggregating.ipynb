```python
# groupby is used to split data into groups based on criteria like categories or values in a column
# aggregating: using functions such as sum, count, mean,... and combine them together
```

```python
import pandas as pd
```

```python
data = {
    'Region':['East', 'West', 'East', 'South', 'North', 'North'],
    'Name':['Alice', 'Bob', 'Charles', 'David', 'Eva', 'Darwin'],
    'Sales':[200,150,300,100,400,100]
}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
# group data by region
grouped = df.groupby('Region')
```

```python
print(grouped)
```

```python
# apply aggregation functions
grouped['Sales'].sum()
```

```python
grouped['Sales'].mean()
```

```python
grouped['Sales'].count()
```

```python
# multiple aggregations
grouped['Sales'].agg(['sum', 'mean', 'count','max', 'min'])
```

```python
# group by multiple column
# đây là code làm nhanh
df.groupby(['Region', 'Sales'])['Sales'].agg(['sum', 'mean', 'count', 'max', 'min'])

# làm chậm thì sẽ như này
'''
grouped_multiple = df.groupby(['Region', 'Sales'])
group_multiple['Sales'].agg(['sum', 'mean', 'count', 'max', 'min'])
'''
```

```python
# convert the result back to a df with the default indexing
grouped['Sales'].sum().reset_index()
```

```python
# apply sorting
grouped['Sales'].sum().sort_values(ascending = False)
```

```python
# value counts and unique values in pandas
# useful in exploring categorical data
```

```python
data = {
    'City':['New York', 'London', 'New York', 'Paris', 'London', 'Berlin', 'Paris', 'Paris'],
    'Gender':['Male', 'Female', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male']
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# value counts - đếm tần suất từng biến
df['City'].value_counts()
```

```python
df['Gender'].value_counts()
```

```python
# get percentage instead of raw count
df['Gender'].value_counts(normalize = True)
```

```python
# unique fuction
df['City'].unique()
```

```python
df['Gender'].unique()
```

```python
# count the number of unique values
df['City'].nunique()
```

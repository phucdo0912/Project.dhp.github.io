```python
import pandas as pd
import numpy as np
```

```python
data = {
    'name': ['Alice', 'Bob', 'Charles', 'David', 'Eva', 'Phuc'],
    'age': ['25', 'np.nan', '30', '29', 'NaN', '25'],
    'gender': ['female','male','male','np.nan','female','female'],
    'country': ['USA', 'usa', 'canada', 'CANADA', 'UK', 'USA'],
    'quantity': [2,1,3,5,2,2],
    'unit_price': [20,15,np.nan,10,30,20]
}

df = pd.DataFrame(data)
df
```

```python
# data cleaning
# handle missing values - mean, median, mode or remove
```

```python
# filling with 0
df.fillna(0)
```

```python
# drop value
df.dropna()
```

```python
# drop duplicate values
df.drop_duplicates()
```

```python
# correcting datatypes
# country
df['country'] = df['country'].str.lower()
df
```

```python
# age - biến cột age thành số nguyên sạch

# chuyển cột age thành kiểu số - không chuyển được thì sẽ biến thành NaN (errors = coerce)
df['age'] = pd.to_numeric(df['age'], errors = 'coerce')

# fill các ô NaN bằng giá trị trung viị của cột
df['age'] = df['age'].fillna(df['age'].median())

# chuyển dữ liệu từ float => int cho phù hợp với tính chất của age
df['age'] = df['age'].astype(int)
```

```python
# data preparation
# make data ready for analysis or modeling
```

```python
# feature engineering - create columns from existing ones
df
```

```python
df['total_price'] = df['unit_price'] * df['quantity']
print(df)
```

```python
# encoding categorial values - converting text labels => specific numbers
pd.get_dummies(df['gender'])
```

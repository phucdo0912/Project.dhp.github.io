```python
# conditionals => used to make decision in code based on certain conditions
# conditions: different criterias

# 3 basic conditional statements if-elif-else
# if runs when the condition is TRUE
# elif runs when you have to check another condition if the first one is FALSE
# else to check if all condition above are FALSE
```

```python
# create variables
x = 10
if x > 10:
  print('positive number')
elif x == 0:
  print('zero')
else:
  print('negative number')

# dấu "=" dùng để gán
# dấu "==" là phép so sánh, kết quả sẽ cho ra là True hoặc False
```

```python
# loops => repeat a block of code multiple times
# for loops - while loops

# for: used to iterate over a sequence like a list or a range of data (lặp đi lặp lại trên một chuỗi)
# while: repeat as long as the condition is TRUE
```

```python
# for loop
for i in range(5):
  print(i)
# range(start, stop) - start: bắt đầu - stop: không bao gồm
# lưu ý: index starts from 0
```

```python
# while loop
count = 0
while count <= 5:
  print(count)
  count = count + 1
  # count += 1
  # có thể dùng dòng này thay cho dòng trên - nhma dễ rối bỏ qua cũng được
```

```python
# for loop
for a in range(1,11):
  if a % 2 == 0:
    print(f"{a} is even")
  else:
    print(f"{a} is odd")
```

```python
# apply loops

correct_number = 7

while True:
  guess = int(input('guess a number between 1 and 10: '))
# tạo ô nhập input với đk input là số nguyên - input: nhận dữ liệu từ người dùng (dạng chuỗi)
  if guess < correct_number:
    print('too low! try again')
  elif guess > correct_number:
    print('too high! try again')
  else:
    print('correct')
    break # thoát vòng lặp
```

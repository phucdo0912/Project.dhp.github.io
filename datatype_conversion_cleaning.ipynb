```python
# data type conversion: proccess of converting data from one type to another
```

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'AGE': ['45','22','30'],
    'Salary':[40000,50000,60000],
    'Join_date':['2025-01-01','2025-07-15', '2025-04-20']
}

df =pd.DataFrame(data)
print(df)
```

```python
df.info()
```

```python
# convert to numeric type
df['AGE'] = pd.to_numeric(df['AGE'])

# convert to int (age is formated in integer)
df['Age'] = df['AGE'].astype(int)
```

```python
# convert to date type
df['Join_date'] = pd.to_datetime(df['Join_date'])
```

```python
print(df.dtypes)
```

```python
# conversion ensures that correst mathematical operations run smoothly
```

```python
# add new column for demonstration
df['Height'] = ['5.8', '6.1','5.5']
df['Department'] = ['HR', 'Finance', 'HR']
df['Experience'] = ['2', 'three', '4']

print(df)
```

```python
# convert to float type
df['Height'] = df['Height'].astype('float')
```

```python
# convert to category
df['Department'] = df['Department'].astype('category')
```

```python
# convert to number with errors
df['Experience'] = pd.to_numeric(df['Experience'], errors = 'coerce')
```

```python
print(df)
```

```python
data = {
    'Name': ['Alice','Bob', None,'David', 'Alice'],
    'Age': ['25','thirty','35', None, '25'],
    'Salary':[50000,60000, None, 55000, 50000],
    'Join_date':['2025-01-01', 'not_availabel', '2025-05-01', '2025-11-01', '2025-01-01']
}

df = pd.DataFrame(data)
df
```

```python
# check for data type
df.info()
```

```python
# summary data
print(df.describe(include = 'all'))
```

```python
# check for missing value
df.isnull().sum()
```

```python
# clean data

# Age
df['Age'] = df['Age'].replace('thirty', np.nan)
df['Age'] = pd.to_numeric(df['Age'], errors = 'coerce')

df['Age'] = df['Age'].fillna(df['Age'].mean())

# Name
df['Name'] = df['Name'].fillna('Unknown')

# Salary
df['Salary'] = df['Salary'].ffill()
```

```python
# convert datatypes

# Join_date
df['Join_date'] = pd.to_datetime(df['Join_date'], errors = 'coerce')
```

```python
# remove duplicates
df = df.drop_duplicates()
```

```python
print(df)
print('Data type \n', df.dtypes)
```

```python
# rename variables
df = df.rename(columns = {'Name': 'First_name', 'Salary': 'income'})
```

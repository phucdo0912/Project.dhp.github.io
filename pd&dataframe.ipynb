```python
# check for null values
# True → ô bị thiếu (NaN)
# False → ô có dữ liệu
df.isnull().sum()
```

```python
# xóa các dòng có giá trị bị thiếu - xóa quan sát
df.dropna(inplace = True) # True là các dòng chứa giá trị bị thiếu phía trên
```

```python
# fil data in missing values
df.fillna(0, inplace = True) # thay thế các giá trị null = 0
```

```python
# export to csv
df.to_csv('output.csv', index = False) # index = False là không ghi cột index ra file csv còn true là sẽ có thêm cột index
```
